```python
# Calculator
import math
def addition(num1, num2):
    return num1 + num2
def subtraction(num1, num2):
    return num1 - num2
def square_root(num):
    return math.sqrt(num)
def division(num1, num2):
    return num1/num2

while True:
    print("\n\nClick 1 for Addition")
    print("Click 2 for Substraction")
    print("Click 3 for SquareRoot")
    print("Click 4 for Division")
    print("Click 0 to Exit")

    try:
        Choice=int(input("Enter any number from 0/1/2/3/4: "))
        if Choice in [1,2]:
            try:
                num1 = float(input("\tEnter first number: "))
                num2 = float(input('\tEnter second number: '))
            except ValueError as VE:
                print(f"\nEnter only digits. Error:{VE}")
                continue
                
            if Choice == 1:
                print(f'\nAddition of {num1} and {num2} is {addition(num1, num2)}')
            elif Choice == 2:
                print(f'\nSubtraction of {num1} and {num2} is {subtraction(num1, num2)}')
                
        elif Choice == 4:
            try:
                num1 = float(input("\tEnter first number: "))
                num2 = float(input('\tEnter second number: '))
                print(f'\nDivision of {num1} and {num2} is {round(division(num1, num2),2)}')
            except ValueError as VE:
                print(f'\nEnter only digits for Division. Error:{VE}')
            except ZeroDivisionError as ZDE:
                print(f"\nDivisor must be non zero. Error:{ZDE}")
                continue # exiting loop after error           
        elif Choice == 3:
            try :
                num = float(input(("\tEnter number to get its square root: ")))
                print(f'square_root of {num} is {round(square_root(num),2)}')
            except ValueError as VE:
                print(f'\nEnter only positive digits. Error:{VE}')
        elif Choice == 0:
            print(f'Exiting from Calculator')
            break
        else:
            print('\nEnter any valid response from 0/1/2/3/4')
    except ValueError as VE:
        print('\nEnter any valid response from 0/1/2/3/4')
    
    
```

```python
# tic tac toe game
import numpy as np
matrix = np.full((3,3),'-')
print("Start your Game by entering index \n ", matrix)

def get_empty_indices(matrix):
    ''' 
    This function returns list of non empty indices.
    '''
    empty_indices = [index for index in np.ndindex(matrix.shape) if matrix[index] == '-']
    empty_indices = [f'{i}{j}' for i,j in empty_indices]
    return empty_indices
    
def winning_condition(matrix):
    '''
    This function checks all possible winning conditions.
    Checks diagnol indices, row indices and column indices,
    If they are having same symbol around any column, row or diagnol
    it returns winning message.
    '''
    winner_msg = ''
    # checking diagnol indices
    if (matrix[0][0] == matrix[1][1] == matrix[2][2] == 'X') or (matrix[0][2] == matrix[1][1] == matrix[2][0] == 'X'):
       winner_msg = "Player 2 with 'X' symbol won"
    elif (matrix[0][0] == matrix[1][1] == matrix[2][2] == 'O') or (matrix[0][2] == matrix[1][1] == matrix[2][0] == 'O'):
        winner_msg = "Player 1 with 'O' symbol won"
        
    # checking all columns
    for col in range(matrix.shape[1]):
        if np.all(matrix[:,col]=='X'):
            winner_msg = "Player 2 with'X' symbol won"
        elif np.all(matrix[:,col]=='O'):
            winner_msg = "Player 1 with 'O' symbol won"
            
     # checking all rows
    for row in matrix:
        if np.all(row == 'X'):
            winner_msg = "Player 2 with'X' symbol won"
        elif np.all(row == 'O'):
            winner_msg = "Player 1 with'O' symbol won"
    return winner_msg


print("To play Tic Tac Toe game, Player has to give indices to place their respective symbols('O' or 'X').")

sample_matrix = np.zeros((3,3), dtype='object')
for i in range(sample_matrix.shape[0]):
    for j in range(sample_matrix.shape[1]):
        sample_matrix[i][j] = f'{i}{j}'
print(f'Player can use any of these indices to play. \n{sample_matrix}')

while True: 
    # Taking input from player1
    player1 = input("Your symbol is 'O' choose index to place.eg:'02'\n")
    
    if player1 in get_empty_indices(matrix):
        i,j = int(player1[0]), int(player1[1])
        matrix[i,j] = 'O'
        print(matrix)   
    elif (player1 not in get_empty_indices(matrix)) or (player1 != 'exit'):
        print(f"index {player1} is not valid or might be filled!.Choose only {get_empty_indices(matrix)}")
        player1 = input("Your symbol is 'O' choose index to place.eg:'02'\n")
        if player1 in get_empty_indices(matrix):
            i,j = int(player1[0]), int(player1[1])
            matrix[i,j] = 'O'
            print(matrix) 
            
    winner_msg = winning_condition(matrix)
    if winner_msg:
        print(f"Winner! {winner_msg}")
        break    
    elif not get_empty_indices(matrix):
        print(f'Game Over!')
        break
    elif player1 == 'exit':
        break
    # Taking input from player2
    player2 = input("Your symbol is 'X' choose index to place.eg:'02'\n")
    
    if player2 in get_empty_indices(matrix):
        i,j = int(player2[0]), int(player2[1])
        matrix[i,j] = 'X'
        print(matrix)
    elif (player2 not in get_empty_indices(matrix)) or (player2 != 'exit'):
        print(f"index {player2} is not valid or might be filled!.Choose only {get_empty_indices(matrix)}")
        player2 = input("Your symbol is 'X' choose index to place.eg:'02'\n")
        if player2 in get_empty_indices(matrix):
            i,j = int(player2[0]), int(player2[1])
            matrix[i,j] = 'X'
            print(matrix)
            
    winner_msg = winning_condition(matrix)
    if winner_msg:
        print(f"Winner! {winner_msg}")
        break
    elif not get_empty_indices(matrix):
        print(f'Game Over! Its Draw.')
        break
    elif player2 == 'exit':
        break
```

```python
print("Choose Number from 1 to 100")
correct = 98
i = 0
while i <=10:
    try:
        chance = int(input('Guess one number, you have 10 chances'))
        i = i + 1
    except ValueError as VE:
        print('Enter valid Number b/w 1 to 100. Error Occured',VE)
        continue
        
    if chance == correct:
        print(chance)
        print('Congratulations! You got it right!')
        break
    elif chance > correct:
        print(chance)
        print('Last Guess was too high!')
        continue
    elif chance < correct:
        print(chance)
        print('Last Guess was too low!')
        continue    
    
```
