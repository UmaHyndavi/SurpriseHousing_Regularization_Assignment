```python
import matplotlib.pyplot as plt
import random

# Toss sizes to simulate
toss_sizes = [10, 50, 100, 200, 300, 400, 500, 1000, 2000]

# Lists to store statistics
means = []
variances = []
ratios = []

for n in toss_sizes:
    # Simulate n coin tosses
    tosses_list = [random.choice(['Heads', 'Tails']) for _ in range(n)]
    
    # Count outcomes
    heads = tosses_list.count('Heads')
    tails = tosses_list.count('Tails')
    
    # Calculate and store ratio of Heads to Tails
    ratio = heads / tails if tails != 0 else float('inf')
    ratios.append(ratio)
    
    # Print empirical probabilities
    print(f'Probabilities for {n} tosses: Heads = {heads/n:.2f}, Tails = {tails/n:.2f}')
    
    # Convert outcomes to numerical values: Heads = 1, Tails = 0
    tosses_numeric = [1 if outcome == 'Heads' else 0 for outcome in tosses_list]
    
    # Compute mean and variance
    mean = sum(tosses_numeric) / n
    means.append(mean)
    
    variance = sum((x - mean) ** 2 for x in tosses_numeric) / n
    variances.append(variance)

# ----------------------------------------
# Plot 1: Ratio of Heads to Tails
# ----------------------------------------

plt.figure(figsize=(10, 6))
plt.plot(toss_sizes, ratios, marker='o', linestyle='-', label='Heads/Tails Ratio')
plt.axhline(y=1.0, color='red', linestyle='--', label='Ideal Ratio = 1.0')
plt.title('Ratio of Heads to Tails as Tosses Increase')
plt.xlabel('Number of Tosses')
plt.ylabel('Heads/Tails Ratio')
plt.legend()
plt.grid(True)
plt.show()

# ----------------------------------------
# Plot 2: Mean and Variance Subplots
# ----------------------------------------

plt.figure(figsize=(14, 6))

# Subplot 1: Mean of Heads
plt.subplot(1, 2, 1)
plt.plot(toss_sizes, means, marker='o', linestyle='-', color='blue', label='Sample Mean')
plt.axhline(y=0.5, color='red', linestyle='--', label='Theoretical Mean = 0.5')
plt.title('Sample Mean of Heads vs Toss Size')
plt.xlabel('Number of Tosses')
plt.ylabel('Mean')
plt.legend()
plt.grid(True)

# Subplot 2: Variance of Outcomes
plt.subplot(1, 2, 2)
plt.plot(toss_sizes, variances, marker='o', linestyle='-', color='green', label='Sample Variance')
plt.axhline(y=0.25, color='red', linestyle='--', label='Theoretical Variance = 0.25')
plt.title('Sample Variance of Toss Outcomes vs Toss Size')
plt.xlabel('Number of Tosses')
plt.ylabel('Variance')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```

##  Answer These Questions

**Q1. What do you expect theoretically?**  
The theoretical probability of getting Heads (or Tails) in a fair coin toss is **0.5**, because both outcomes are equally likely.

---

**Q2. How close is your simulation to theory?**  
As the number of tosses increases, the simulated (empirical) probability gets **closer to 0.5**. This shows that our simulation aligns well with theoretical expectations over larger samples.

---

**Q3. Why do results vary each time?**  
Because of randomness. Especially for **smaller toss sizes**, the variation is higher — it’s easier to get 7 Heads out of 10 than 700 out of 1000. With **more tosses**, the results become more stable due to the **Law of Large Numbers**.

---

**Q4. What happens when you increase tosses? Why?**  
The results **approach the theoretical probability (0.5)**. This happens because increasing tosses **reduces the impact of random fluctuations**, allowing the **true probability to emerge more clearly**.
